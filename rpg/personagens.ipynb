```python
class Personagens:
    def __init__(self, raca):
        self.nivel = 0
        self.xp = 0

        self.vida_atual = 0

        self.forca = 0
        self.destreza = 0
        self.constituicao = 0
        self.inteligencia = 0
        
        self.classe = 0
        self.raca = raca
        
        if raca == 'Anão':
            self.constituicao += 2 + dados(20,1)
            self.forca += 1 + dados(20,1)
            self.inteligencia += 1 + dados(20,1)
            self.destreza += dados(20,1)
        elif raca == 'Elfo':
            self.destreza += 2 + dados(20,1)
            self.inteligencia += 1 + dados(20,1)
            self.forca += dados(20,1)
            self.constituicao += dados(20,1)
        elif raca == 'Halfling':
            self.destreza += 2 + dados(20,1)
            self.constituicao += 1 + dados(20,1)
            self.forca += dados(20,1)
            self.inteligencia += dados(20,1)
        elif raca == 'Humano':
            self.forca += 1 + dados(20,1)
            self.destreza += 1 + dados(20,1)
            self.constituicao += 1 + dados(20,1)
            self.inteligencia += 1 + dados(20,1)
        elif raca == 'Meio-Orc':
            self.forca += 2 + dados(20,1)
            self.constituicao += 1 + dados(20,1)
            self.destreza += dados(20,1)
            self.inteligencia += dados(20,1)
        elif raca == 'Meio-Elfo':
            self.inteligencia += 1 + dados(20,1)
            self.destreza += 1 + dados(20,1)
            self.constituicao += 1 + dados(20,1)
            self.forca += dados(20,1)

    def modificadores_atributos(self):
        pass

    def hp_loss(self, dano):
        self.vida_atual -= dano
        print(f'O {self.classe} {self.raca} tomou {dano} de dano e agora está com {self.vida_atual} de vida!')

    def atacar(self):
        pass

    def upar(self):
        '''Upa o personagem conforme ele ganha XP, ao upar a vida dobra, se cura'''
        tabela_dnd = [
            (355000, 20), (305000, 19), (265000, 18), (225000, 17), (195000, 16),
            (165000, 15), (140000, 14), (120000, 13), (100000, 12), (85000, 11),
            (64000, 10), (48000, 9),  (34000, 8),   (23000, 7),   (14000, 6),
            (6500, 5),    (2700, 4),   (900, 3),    (300, 2),     (0, 1)
        ]
        for xp_total, xp_lvl in tabela_dnd:
            if self.xp >= xp_total:
                if self.nivel < xp_lvl:
                    print(f'Parabens, o {self.classe} {self.raca} upou! indo do nivel {self.nivel}')
                    self.nivel = xp_lvl
                    print(f'para o nível {self.nivel}')
                    self.vida_max += self.vida_max
                    self.vida_atual = self.vida_max
                else: pass

```

```python
cleber = Personagens('Anão')
cleber.xp=23000

print(cleber.forca,
cleber.destreza,
cleber.constituicao,
cleber.inteligencia,
cleber.nivel,
cleber.xp,
cleber.raca,
cleber.vida_max)

cleber.upar()

print(cleber.nivel, cleber.vida_max)
```
